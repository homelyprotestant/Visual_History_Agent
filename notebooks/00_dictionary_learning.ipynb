## Train dictionary

Progress bars cover pipeline stages, Approximate K-SVD iterations, and full-corpus sparse coding.

The dictionary was initialized from 1,000 randomly selected, normalized corpus embeddings. Learning alternates between sparse encoding and dictionary updating. At each iteration, 10% of the corpus was sampled without replacement and encoded against the current dictionary using non-negative Elastic Net. 


```python
from train_global_dictionary import train_global_dictionary

meta = train_global_dictionary(
    project_root=ROOT,
    n_atoms=1000,
    elastic_net_alpha=3e-5,
    l1_ratio=0.5,
    max_iter=11,
    seed=42,
)
print("Dictionary:", artifact_paths(ROOT)["dictionary"])
print("Mean active atoms:", meta["mean_active_atoms"])
```
