# Visual History Agent — 04 Local Dai spatial atom browser

1. Global nonnegative Elastic Net on pooled Z → active atoms + importance  
2. Centered patch-Z **fixed-dictionary NMF** codes against those atoms → LR maps  
3. Local Dai (RGB K-SVD + **OMP**) upsamples to 224×224  

Supports a **two-image comparison** browser and a **single-image** browser. Comparison maps are pair-scaled by Elastic Net importance (p99 normalize).


## Two-image browser (Norton vs Princeton)


```python
from __future__ import annotations

from pathlib import Path
import os
import sys
import importlib

ROOT = Path("../").resolve()
os.environ["VISUAL_HISTORY_AGENT_ROOT"] = str(ROOT)
sys.path.insert(0, str(ROOT))

import visual_history_agent.encode as _enc
import visual_history_agent.models.classical_ksvd as _ksvd
import visual_history_agent.models.local_dai_sr as _dai
import visual_history_agent.spatial_browser as _sb

importlib.reload(_enc)
importlib.reload(_ksvd)
importlib.reload(_dai)
importlib.reload(_sb)

from visual_history_agent.spatial_browser import (
    prepare_spatial_browser,
    display_spatial_atom_browser,
)
from visual_history_agent.paths import artifact_paths

ex = artifact_paths(ROOT)["examples"]
IMAGE_A = ex / "Norton_Jerome.jpg"
IMAGE_B = ex / "Princeton_Jerome.jpg"
FORCE_RECOMPUTE = False

bundle = prepare_spatial_browser(
    IMAGE_A,
    IMAGE_B,
    force_recompute=FORCE_RECOMPUTE,
    device="mps",
)
print("Image A:", IMAGE_A)
print("Image B:", IMAGE_B)
print("Output:", bundle.output_dir)
print("Browser atoms:", len(bundle.browser_atom_ids))
```

```python
display_spatial_atom_browser(bundle)
```

## Single-image browser
